```python
import rasterio
import numpy as np

file_tif = r"C:\Users\LENOVO\PSD\pertemuan_6\a\3.tiff"

with rasterio.open(file_tif) as src:
    band = src.read(1, masked=True)
    valid = band.compressed()

    print("Ukuran:", src.width, "x", src.height)
    print("CRS:", src.crs)
    print("Tipe data:", src.dtypes[0])
    print("Jumlah piksel valid:", valid.size)

    if valid.size:
        print("Minimum:", valid.min())
        print("Maksimum:", valid.max())
        print("Jumlah nilai unik:", np.unique(valid).size)
```
