```python
import rasterio
from rasterio.merge import merge
import glob
import os

# 1. Folder lokasi file Anda
folder_path = r'C:\Users\LENOVO\PSD\pertemuan_6\jawaTimur'

# Ambil semua file TIFF di folder tersebut
semua_file = glob.glob(os.path.join(folder_path, '*.tif*'))

# 2. Mengelompokkan file berdasarkan awalan namanya (contoh: 'B02', 'B03', 'B04', dst)
kelompok_band = {}

for path_file in semua_file:
    nama_file = os.path.basename(path_file)
    
    # Memotong nama file berdasarkan underscore '_'
    # Misalnya 'B02_kiri.tiff' -> menjadi 'B02'
    prefix_band = nama_file.split('_')[0] 
    
    if prefix_band not in kelompok_band:
        kelompok_band[prefix_band] = []
    
    kelompok_band[prefix_band].append(path_file)

# 3. Mulai perulangan untuk menggabungkan masing-masing band
for band, list_file in kelompok_band.items():
    print(f"\nMemproses penggabungan untuk: {band} ({len(list_file)} file ditemukan)...")
    
    # Lewati jika file cuma ada 1 (tidak ada pasangan untuk digabung)
    if len(list_file) < 2:
        print(f"Melewati {band} karena file kurang dari 2.")
        continue
        
    # Buka semua file untuk band spesifik ini
    src_files = [rasterio.open(f) for f in list_file]
    
    try:
        # Lakukan penggabungan
        mosaic, out_trans = merge(src_files)
        
        # Salin dan perbarui metadata
        out_meta = src_files[0].meta.copy()
        out_meta.update({
            "driver": "GTiff",
            "height": mosaic.shape[1],
            "width": mosaic.shape[2],
            "transform": out_trans
        })
        
        # 4. Simpan dengan nama spesifik per band di folder yang sama
        nama_output = f"{band}_gabungan.tiff"
        path_output = os.path.join(folder_path, nama_output)
        
        with rasterio.open(path_output, "w", **out_meta) as dest:
            dest.write(mosaic)
            
        print(f"Berhasil! File disimpan di: {path_output}")
        
    except Exception as e:
        print(f"Gagal memproses {band}. Error: {e}")
        
    finally:
        # Tutup file
        for src in src_files:
            src.close()
```
